OpenCV 與 Matplotlib —— 色彩空間轉換與 RGB 直方圖分析（Color Histogram）

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("RedCat.jpg")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 10))
plt.imshow(image)
plt.axis("off")
plt.title("Original Image")
plt.show()

colors = ('red', 'green', 'blue')

for i, color in enumerate(colors):
  histr = cv2.calcHist([image], [i], None, [256], [0, 256])
  plt.plot(histr, color=color, label=f'{color} channel')
  plt.title('RGB Image Histogram')
  plt.xlabel('Pixel Value')
  plt.ylabel('Frequency')
  plt.legend()
  plt.show()

```

OpenCV 基礎 —— 灰階、負片（反相）與二值化

```python
# 1. 匯入所需函式庫
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 2. 讀取原始圖片 (BGR 格式)
img_bgr = cv2.imread("RedCat.jpg")

# 防呆檢查：確認圖片是否正確讀入
if img_bgr is None:
    print("錯誤：找不到圖片，請確認 'RedCat.jpg' 是否已上傳至目錄！")
else:
    # 3. 影像色彩轉換與前處理
    # 3.1 轉為 RGB 供 Matplotlib 正常顯示彩色
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

    # 3.2 轉為灰階影像 (二維矩陣)
    img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)

    # 3.3 計算反相（負片/反白效果）
    img_inverted = 255 - img_gray

    # 3.4 進行固定閾值二值化 (門檻值設為 127)
    threshold_value, img_binary = cv2.threshold(img_gray, 127, 255, cv2.THRESH_BINARY)

    # 4. 依序繪圖展示
    # 4.1 原始彩色圖
    plt.imshow(img_rgb)
    plt.title("原始圖片 (RGB)")
    plt.axis('off')
    plt.show()

    # 4.2 灰階圖 (注意：需搭配 cmap='gray')
    plt.imshow(img_gray, cmap='gray')
    plt.title("灰階圖片")
    plt.axis('off')
    plt.show()

    # 4.3 反白/負片效果
    plt.imshow(img_inverted, cmap='gray')
    plt.title("反白/負片效果")
    plt.axis('off')
    plt.show()

    # 4.4 二值化黑白圖
    plt.imshow(img_binary, cmap='gray')
    plt.title("二值化效果 (Thresholding)")
    plt.axis('off')
    plt.show()
```

OpenCV 基礎 —— 均勻雜訊模型（Uniform Noise）





```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
from numpy.random import uniform
from google.colab.patches import cv2_imshow

def uniform_noise(f, scale):
    g = f.copy()
    nr, nc = f.shape[:2]
    for x in range(nr):
        for y in range(nc):
            value = f[x, y] + uniform(0, 1) * scale
            g[x, y] = np.uint8(np.clip(value, 0, 255))
    return g

def PSNR(f, g):
    mse = np.mean((f.astype(np.float64) - g.astype(np.float64)) ** 2)
    if mse == 0:
        return float('inf')
    return 10 * np.log10((255.0 ** 2) / mse)

def histogram(f):
    if f.ndim != 3:
        hist = cv2.calcHist([f], [0], None, [256], [0, 256])
        plt.plot(hist)
    else:
        color = ('b', 'g', 'r')
        for i, col in enumerate(color):
            hist = cv2.calcHist([f], [i], None, [256], [0, 256])
            plt.plot(hist, color=col)
    plt.xlim([0, 256])
    plt.xlabel("Intensity")
    plt.ylabel("#Intensities")
    plt.title("Uniform Noise Histogram")
    plt.show()

# --- 主程式 ---
scale = 30  # 雜訊強度範圍
img1 = cv2.imread("RedCat.jpg", -1)

if img1 is None:
    print("找不到圖片，請確認檔名！")
else:
    img2 = uniform_noise(img1, scale)
    print("PSNR =", PSNR(img1, img2))

    print("【原圖】")
    cv2_imshow(img1)
    print("【均勻雜訊圖】")
    cv2_imshow(img2)

    histogram(img2)
```

高斯雜訊模型（Gaussian Noise）




```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
from numpy.random import normal
from google.colab.patches import cv2_imshow

def gaussian_noise(f, scale):
    g = f.copy()
    nr, nc = f.shape[:2]
    for x in range(nr):
        for y in range(nc):
            value = f[x, y] + normal(0, scale)
            g[x, y] = np.uint8(np.clip(value, 0, 255))
    return g

def PSNR(f, g):
    mse = np.mean((f.astype(np.float64) - g.astype(np.float64)) ** 2)
    if mse == 0:
        return float('inf')
    return 10 * np.log10((255.0 ** 2) / mse)

def histogram(f):
    if f.ndim != 3:
        hist = cv2.calcHist([f], [0], None, [256], [0, 256])
        plt.plot(hist)
    else:
        color = ('b', 'g', 'r')
        for i, col in enumerate(color):
            hist = cv2.calcHist([f], [i], None, [256], [0, 256])
            plt.plot(hist, color=col)
    plt.xlim([0, 256])
    plt.xlabel("Intensity")
    plt.ylabel("#Intensities")
    plt.title("Gaussian Noise Histogram")
    plt.show()

# --- 主程式 ---
scale = 25  # 常態分佈的標準差 (sigma)
img1 = cv2.imread("RedCat.jpg", -1)

if img1 is None:
    print("找不到圖片，請確認檔名！")
else:
    img2 = gaussian_noise(img1, scale)
    print("PSNR =", PSNR(img1, img2))

    print("【原圖】")
    cv2_imshow(img1)
    print("【高斯雜訊圖】")
    cv2_imshow(img2)

    histogram(img2)
```

指數雜訊模型（Exponential Noise）

胡椒鹽雜訊模型（Salt and Pepper Noise）

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
from numpy.random import uniform
from google.colab.patches import cv2_imshow

def salt_pepper_noise(f, probability):
    g = f.copy()
    nr, nc = f.shape[:2]
    for x in range(nr):
        for y in range(nc):
            value = uniform(0, 1)
            if value <= probability / 2:
                g[x, y] = 0        # 胡椒 (Pepper): 全黑
            elif value <= probability:
                g[x, y] = 255      # 鹽粒 (Salt): 全白
            else:
                g[x, y] = f[x, y]  # 保持原像素
    return g

def PSNR(f, g):
    mse = np.mean((f.astype(np.float64) - g.astype(np.float64)) ** 2)
    if mse == 0:
        return float('inf')
    return 10 * np.log10((255.0 ** 2) / mse)

def histogram(f):
    if f.ndim != 3:
        hist = cv2.calcHist([f], [0], None, [256], [0, 256])
        plt.plot(hist)
    else:
        color = ('b', 'g', 'r')
        for i, col in enumerate(color):
            hist = cv2.calcHist([f], [i], None, [256], [0, 256])
            plt.plot(hist, color=col)
    plt.xlim([0, 256])
    plt.xlabel("Intensity")
    plt.ylabel("#Intensities")
    plt.title("Salt and Pepper Noise Histogram")
    plt.show()

# --- 主程式 ---
probability = 0.05  # 5% 的像素受到雜訊污染
img1 = cv2.imread("RedCat.jpg", -1)

if img1 is None:
    print("找不到圖片，請確認檔名！")
else:
    img2 = salt_pepper_noise(img1, probability)
    print("PSNR =", PSNR(img1, img2))

    print("【原圖】")
    cv2_imshow(img1)
    print("【胡椒鹽雜訊圖】")
    cv2_imshow(img2)

    histogram(img2)
```
